```python
from ultralytics import YOLO
```

```python
import cv2 
import matplotlib.pyplot as plt
model = YOLO("best.pt")
result = model("test1.jpg" , show = True)
res_plotted = result[0].plot()

plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(res_plotted, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
